```python
input_folder = Path(
    r"C:\Users\dawit\PycharmProjects\hydration break\data\2026"
)
output_folder = Path(
    r"C:\Users\dawit\PycharmProjects\hydration break\data\2026_cleaned"
)
output_folder.mkdir(parents=True, exist_ok=True)
# Process all 2026 matches
csv_files = sorted(input_folder.glob("*.csv"))
print(f"Found {len(csv_files)} CSV files.")
for file in csv_files:
    print(f"Processing: {file.name}")
    match_2026 = pd.read_csv(file)
    column_n = [
        "event",
        "match_id",
        "minute",
        "second",
        "period",
        "team",
        "outcome",
        "qualifiers",
        "isShot"
    ]
    match_2026_clean = match_2026[column_n].copy()
    # 2. Drop qualifiers
    match_2026_clean = match_2026_clean.drop(
        columns="qualifiers",errors="ignore"
    )
    # 3. One-hot encode event
    match_2026_dummies = pd.get_dummies(
        match_2026_clean,
        columns=["event"],
        dtype=int
    )
    # 4. Remove irrelevant event dummy columns
    column_to_drop = [
        "event_End",
        "event_Start",
        "event_FormationSet",
        "event_FormationChange",
        "event_SubstitutionOff",
        "event_SubstitutionOn",
        "event_KeeperPickup",
        "event_KeeperSweeper",
        "event_Save",
        "event_Smother",
        "event_Error",
        "event_OffsideGiven",
        "event_OffsidePass",
        "event_OffsideProvoked",
        "event_Card",
        "event_Goal"
    ]
    match_2026_dummies = match_2026_dummies.drop(
        columns=column_to_drop,errors="ignore"
    )
    # 5. Convert period into simple period names
    match_2026_dummies["period"] = (
        match_2026_dummies["period"]
        .apply(ast.literal_eval)
        .apply(lambda d: d["displayName"])
    )
    # 6. Keep only FirstHalf and SecondHalf
    period = ["FirstHalf", "SecondHalf"]
    match_2026_dummies = match_2026_dummies[
        match_2026_dummies["period"].isin(period)
    ]
    # 7. Convert isShot to 0/1
    match_2026_dummies = match_2026_dummies.assign(
        isShot=lambda x: np.where(
            x["isShot"] == True,
            1,
            0
        )
    )
    # 8. Rename the nine common features
    match_2026_dummies = match_2026_dummies.rename(
        columns={
            "event_BallRecovery": "ball_recovery",
            "event_Clearance": "clearance",
            "event_Foul": "foul",
            "event_Pass": "pass",
            "event_Tackle": "tackle",
            "event_Dispossessed": "dispossessed",
            "isShot": "shot",
            "event_Interception": "interception",
            "event_TakeOn": "dribble"
        }
    )
    # 9. Create match_time
    match_2026_dummies["match_time"] = (
        pd.to_timedelta(
            match_2026_dummies["minute"],
            unit="m"
        )
        +
        pd.to_timedelta(
            match_2026_dummies["second"].fillna(0),
            unit="s"
        )
    )
    # 10. Drop minute and second
    match_2026_dummies = match_2026_dummies.drop(
        columns=["minute", "second"]
    )
    # 11. Keep final analysis columns
    column_not_to_drop = [
        "match_id",
        "period",
        "shot",
        "team",
        "ball_recovery",
        "clearance",
        "dispossessed",
        "foul",
        "interception",
        "dribble",
        "match_time",
        "outcomeType",
        "pass",
        "tackle"
    ]
    for column in match_2026_dummies.columns:
        if column not in column_not_to_drop:
            match_2026_dummies = match_2026_dummies.drop(
                columns=column,errors="ignore"
            )
    # 12. Extract the four analysis windows
    (
        match_2026_12_22,
        match_2026_25_35,
        match_2026_57_67,
        match_2026_70_80
    ) = time_extractor(
        match_2026_dummies,
        2026
    )
    # 13. Save the four cleaned datasets
match_id = str(match_2026["match_id"].iloc[0])
output_file_1 = output_folder / f"{match_id}_2026_12_22.csv"
output_file_2 = output_folder / f"{match_id}_2026_25_35.csv"
output_file_3 = output_folder / f"{match_id}_2026_57_67.csv"
output_file_4 = output_folder / f"{match_id}_2026_70_80.csv"
if output_file_1.exists():
    match_2026_12_22.to_csv(
        output_file_1,
        index=False
    )
if output_file_2.exists():
    match_2026_25_35.to_csv(
        output_file_2,
        index=False
    )
if output_file_3.exists():
    match_2026_57_67.to_csv(
        output_file_3,
        index=False
    )
if output_file_4.exists():
    match_2026_70_80.to_csv(
        output_file_4,
        index=False
    )
```
