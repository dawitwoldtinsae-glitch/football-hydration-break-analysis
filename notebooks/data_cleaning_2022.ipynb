```python
input_folder = Path(
    r"C:\Users\dawit\PycharmProjects\hydration break\data\2022"
)
output_folder = Path(
    r"C:\Users\dawit\PycharmProjects\hydration break\data\2022_cleaned"
)
output_folder.mkdir(parents=True, exist_ok=True)
for file in input_folder.glob("*.json"):
    data = pd.read_json(file)
    match_id = int(file.stem)
    # Step 1: Extract event name from type
    data = data.assign(
        type=lambda x: x['type'].apply(lambda d: d['name'])
    )
    # Step 2: Replace Duel with its specific subtype
    data['type'] = data.apply(
        lambda row: row['type']
        if pd.isna(row['duel'])
        else row['duel']['type']['name'],
        axis=1
    )
    # Step 3: Keep required columns
    columns = [
        'team',
        'minute',
        'second',
        'type',
        'period',
        'play_pattern'
    ]
    for column in data.columns:
        if column not in columns:
            data = data.drop(columns=column, errors='ignore')
    # Step 4: Extract names and add match ID
    data = data.assign(
        play_pattern=lambda x: x['play_pattern'].apply(lambda d: d['name']),
        team=lambda x: x['team'].apply(lambda d: d['name']),
        match_id=match_id
    )
    # Step 5: Create match_time
    data['match_time'] = (
        pd.to_timedelta(data['minute'], unit='m')
        + pd.to_timedelta(data['second'], unit='s')
    )
    # Step 6: Keep the nine common event types
    column_not_drop = [
        'Pass',
        'Clearance',
        'Ball Recovery',
        'Dispossessed',
        'Foul Committed',
        'Shot',
        'Interception',
        'Dribble',
        'Tackle'
    ]
    data_clean = data[
        data['type'].isin(column_not_drop)
    ]
    # Step 7: One-hot encode event types
    data_dummy = pd.get_dummies(
        data_clean,
        columns=['type'],
        dtype=int
    )
    # Step 8: Rename event columns
    data_dummy = data_dummy.rename(
        columns={
            'type_Pass': 'pass',
            'type_Shot': 'shot',
            'type_Foul Committed': 'foul',
            'type_Interception': 'interception',
            'type_Dribble': 'dribble',
            'type_Clearance': 'clearance',
            'type_Ball Recovery': 'ball_recovery',
            'type_Dispossessed': 'dispossessed',
            'type_Tackle': 'tackle'
        }
    )
    # Step 9: Extract the four 2022 time windows
    data_12_22, data_22_32, data_57_67, data_67_77 = (
        time_extractor(data_dummy, 2022)
    )
    # Save each window as CSV
    windows = {
        "12_22": data_12_22,
        "22_32": data_22_32,
        "57_67": data_57_67,
        "67_77": data_67_77
    }
    for window_name, window_data in windows.items():

        output_file = output_folder / (
            f"{match_id}_2022_{window_name}.csv"
        )
        # to_csv overwrites an existing file
        window_data.to_csv(
            output_file,
            index=False
        )
print("2022 cleaning and extraction complete.")
```
